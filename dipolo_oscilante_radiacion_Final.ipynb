## **Problema: Radiación de un dipolo eléctrico oscilante** ##

---

**Enunciado.** Un dipolo eléctrico puntual en el origen oscila armónicamente en la dirección $z$:

$$\mathbf p(t)=p_0\cos(\omega t)\,\hat{\mathbf z}$$

Se solicita construir un programa que visualice la estructura espacial de $\mathbf E(\mathbf r,t)$ y $\mathbf B(\mathbf r,t)$ a un tiempo $t$ dado, explorando distintas distancias respecto a la longitud de onda

$$\lambda=\frac{2\pi c}{\omega}$$

para comparar la **región cercana** con la **región de radiación**. Además se pide:

- identificar los términos de los campos con distinta dependencia en $r$ y analizar cuáles dominan si $r\ll\lambda$ y cuáles sobreviven si $r\gg\lambda$;
- calcular el vector de Poynting $\mathbf S=\frac1{\mu_0}\mathbf E\times\mathbf B$ y su promedio temporal $\langle\mathbf S\rangle=\frac1T\int_0^T\mathbf S\,dt$, y usarlos para estudiar el flujo de energía;
- determinar y representar, en la región de radiación, la distribución angular de la potencia radiada $dP/d\Omega$, y relacionarla con la orientación del momento dipolar.


---
## **Solución** ##


### **1. Planteamiento y aproximaciones**

Se modela el dipolo como dos cargas $\pm q(t)$ ubicadas en $z=\pm d/2$, unidas por un hilo por el que circula la corriente necesaria para cambiar las cargas, con

$$q(t)=q_0\cos\omega t,\qquad p_0=q_0\,d$$

Aproximaciones:

- $d\ll r$: el observador está lejos comparado con el tamaño del dipolo.
- $d\ll\lambda$: equivale a $\omega d/c\ll1$, es decir, el dipolo es pequeño comparado con la longitud de onda.

**No se requiere $r\gg\lambda$.** Eso es lo que permite estudiar la región cercana ($r\lesssim\lambda$) y la de radiación ($r\gg\lambda$) con las *mismas* fórmulas, y ver cómo una se convierte en la otra.

Notación: $k=\omega/c=2\pi/\lambda$, $\;u\equiv\omega(t-r/c)=\omega t-kr$ (fase retardada), $\;\xi\equiv kr=2\pi r/\lambda$.

**Interpretación física.** El parámetro $\xi=kr$ es la distancia medida en unidades de $\lambda/2\pi$. Es la única combinación adimensional que controla la forma de los campos: $\xi\ll1$ es zona cercana y $\xi\gg1$ es zona de radiación.

---

### **2. Potencial escalar retardado**

$$V(\mathbf r,t)=\frac{q_0}{4\pi\epsilon_0}\left[\frac{\cos\omega(t-\mathfrak r_+/c)}{\mathfrak r_+}-\frac{\cos\omega(t-\mathfrak r_-/c)}{\mathfrak r_-}\right]$$

donde $\mathfrak r_\pm$ es la distancia de la carga $\pm q$ al punto de observación.

**Interpretación física.** El potencial en $\mathbf r$ en el instante $t$ depende de lo que hacían las cargas en el instante *retardado* $t-\mathfrak r_\pm/c$, porque la información viaja a velocidad finita $c$.

**Paso 1: distancias.** Por la ley de cosenos y $d\ll r$:

$$\mathfrak r_\pm=\sqrt{r^2\mp rd\cos\theta+d^2/4}\simeq r\mp\frac d2\cos\theta,\qquad \frac1{\mathfrak r_\pm}\simeq\frac1r\left(1\pm\frac{d\cos\theta}{2r}\right)$$

**Paso 2: argumento del coseno.** Con $u=\omega t-kr$:

$$\omega\left(t-\frac{\mathfrak r_\pm}{c}\right)\simeq u\pm\frac{kd\cos\theta}{2}$$

y como $kd\ll1$ se expande a primer orden:

$$\cos\left(u\pm\frac{kd\cos\theta}{2}\right)\simeq\cos u\mp\frac{kd\cos\theta}{2}\sin u$$

**Paso 3: sustitución.** Definiendo $\delta=\dfrac{d\cos\theta}{2r}$ y $\varepsilon=\dfrac{kd\cos\theta}{2}$:

$$V=\frac{q_0}{4\pi\epsilon_0r}\Big[(1+\delta)(\cos u-\varepsilon\sin u)-(1-\delta)(\cos u+\varepsilon\sin u)\Big]=\frac{q_0}{4\pi\epsilon_0r}\Big[2\delta\cos u-2\varepsilon\sin u\Big]$$

(se descartan los productos $\delta\varepsilon$, de segundo orden). Con $q_0d=p_0$:

$$\boxed{V(r,\theta,t)=\frac{p_0\cos\theta}{4\pi\epsilon_0\,r}\left[\frac{\cos u}{r}-k\sin u\right]}$$

**Interpretación física.** El primer término $\cos u/r^2$ es el potencial del dipolo electrostático, pero evaluado con retardo. El segundo, $k\sin u/r$, no existe en el caso estático, aparece por el retardo y decae solo como $1/r$: es la semilla de la radiación.

---

### **3. Potencial vectorial retardado**

La corriente en el dipolo es $I=\dot q=-q_0\omega\sin\omega t$ en dirección $+\hat z$, y $\int I\,dl=\dot p\,\hat z$ con $\dot p=-p_0\omega\sin\omega t$. Como $d\ll r$ se toma $\mathfrak r\approx r$:

$$\boxed{\mathbf A(r,t)=-\frac{\mu_0p_0\omega}{4\pi r}\sin u\;\hat{\mathbf z}}$$

Para derivar en coordenadas esféricas se usa $\hat z=\cos\theta\,\hat r-\sin\theta\,\hat\theta$. Llamando $A\equiv-\dfrac{\mu_0p_0\omega}{4\pi}\dfrac{\sin u}{r}$:

$$A_r=A\cos\theta,\qquad A_\theta=-A\sin\theta,\qquad A_\phi=0$$

**Interpretación física.** $\mathbf A$ sigue a la corriente $\dot p$, que está desfasada $90^\circ$ respecto a $p$. Esto ya anticipa que el campo magnético de zona cercana estará desfasado $90^\circ$ respecto al eléctrico.

---

### **4. Campo magnético: $\mathbf B=\nabla\times\mathbf A$**

Por simetría azimutal ($\partial_\phi=0$) y $A_\phi=0$, solo sobrevive la componente $\phi$:

$$B_\phi=\frac1r\left[\frac{\partial(rA_\theta)}{\partial r}-\frac{\partial A_r}{\partial\theta}\right]$$

**Primer término:** $rA_\theta=\dfrac{\mu_0p_0\omega}{4\pi}\sin\theta\,\sin u$. Como $\partial u/\partial r=-k$:

$$\frac{\partial(rA_\theta)}{\partial r}=-\frac{\mu_0p_0\omega}{4\pi}\,k\sin\theta\cos u$$

**Segundo término:** $\dfrac{\partial A_r}{\partial\theta}=-A\sin\theta=\dfrac{\mu_0p_0\omega}{4\pi}\dfrac{\sin\theta\sin u}{r}$.

**Resultado:**

$$B_\phi=\frac1r\cdot\frac{\mu_0p_0\omega}{4\pi}\sin\theta\left[-k\cos u-\frac{\sin u}{r}\right]$$

$$\boxed{B_\phi=-\frac{\mu_0p_0\omega k\sin\theta}{4\pi r}\left[\cos u+\frac{\sin u}{kr}\right]}$$

**Comprobación (zona cercana, $kr\ll1$).**

Domina el segundo término:

$$B_\phi\to-\frac{\mu_0p_0\omega\sin\theta}{4\pi r^2}\sin u=\frac{\mu_0}{4\pi}\frac{\dot p\,\sin\theta}{r^2},$$

que es la ley de Biot–Savart para un elemento de corriente $I\,dl=\dot p$.

**Interpretación física.** El término $\propto1/r$ (con $\cos u$) es el campo de radiación. El término $\propto1/r^2$ (con $\sin u$) es el análogo de Biot–Savart para la corriente $\dot p$ y está en cuadratura con el otro. Las líneas de $\mathbf B$ son círculos alrededor del eje $z$.

---

### **5. Campo eléctrico: $\mathbf E=-\nabla V-\partial_t\mathbf A$**

Se usa $\mu_0\omega^2=k^2/\epsilon_0$ (porque $\mu_0\epsilon_0=1/c^2$). Se abrevia $C\equiv\dfrac{p_0}{4\pi\epsilon_0}$, de modo que $V=C\cos\theta\left[\dfrac{\cos u}{r^2}-\dfrac{k\sin u}{r}\right]$.

1. **Componente radial**

Paso 1: derivadas respecto a $r$.

Como $\partial u/\partial r=-k$, se cumple $\partial_r\cos u=k\sin u$ y $\partial_r\sin u=-k\cos u$. Entonces

$$\frac{\partial}{\partial r}\left(\frac{\cos u}{r^2}\right)=\frac{k\sin u}{r^2}-\frac{2\cos u}{r^3}$$

$$\frac{\partial}{\partial r}\left(-\frac{k\sin u}{r}\right)=-k\left[\frac{-k\cos u}{r}-\frac{\sin u}{r^2}\right]=\frac{k^2\cos u}{r}+\frac{k\sin u}{r^2}$$

Sumando:

$$-\frac{\partial V}{\partial r}=-C\cos\theta\left[\frac{k^2\cos u}{r}+\frac{2k\sin u}{r^2}-\frac{2\cos u}{r^3}\right]$$

Paso 2: derivada temporal de $A_r$.

Con $A_r=A\cos\theta$ y $\partial_tA=-\dfrac{\mu_0p_0\omega^2}{4\pi}\dfrac{\cos u}{r}$:

$$-\frac{\partial A_r}{\partial t}=+\frac{\mu_0p_0\omega^2}{4\pi}\frac{\cos\theta\cos u}{r}=Ck^2\cos\theta\,\frac{\cos u}{r}$$

Paso 3: suma.

El término $k^2\cos u/r$ **se cancela**:

$$E_r=\frac{p_0\cos\theta}{4\pi\epsilon_0}\,\frac{2}{r^3}\Big[\cos u-kr\sin u\Big]$$

2. **Componente polar**

Paso 1: derivada respecto a $\theta$.

Como $\partial_\theta\cos\theta=-\sin\theta$:

$$-\frac1r\frac{\partial V}{\partial\theta}=\frac{C\sin\theta}{r}\left[\frac{\cos u}{r^2}-\frac{k\sin u}{r}\right]=C\sin\theta\left[\frac{\cos u}{r^3}-\frac{k\sin u}{r^2}\right]$$

Paso 2: derivada temporal de $A_\theta$.

Con $A_\theta=-A\sin\theta$:

$$-\frac{\partial A_\theta}{\partial t}=+\sin\theta\,\partial_tA=-Ck^2\sin\theta\,\frac{\cos u}{r}$$

Paso 3: suma

$$E_\theta=C\sin\theta\left[\frac{\cos u}{r^3}-\frac{k\sin u}{r^2}-\frac{k^2\cos u}{r}\right]=\frac{p_0\sin\theta}{4\pi\epsilon_0r^3}\Big[(1-k^2r^2)\cos u-kr\sin u\Big]$$

---

**Comprobación (límite estático, $k\to0$).** $E_r\to\dfrac{2p_0\cos\theta}{4\pi\epsilon_0r^3}$ y $E_\theta\to\dfrac{p_0\sin\theta}{4\pi\epsilon_0r^3}$, el campo del dipolo electrostático.

---

**Comprobación (zona lejana, $kr\gg1$).** Solo sobrevive el término $\propto1/r$ de $E_\theta$:

$$E_\theta\to-\frac{p_0k^2\sin\theta}{4\pi\epsilon_0r}\cos u=-\frac{\mu_0p_0\omega^2\sin\theta}{4\pi r}\cos u$$

y, del campo magnético de la sección 4, usando $\omega=ck$,

$$cB_\phi\to-\frac{c\,\mu_0p_0\omega k\sin\theta}{4\pi r}\cos u=-\frac{\mu_0p_0\omega^2\sin\theta}{4\pi r}\cos u\;=\;E_\theta .$$

Es decir, $E_\theta=cB_\phi$ en módulo y fase: onda transversal, como se espera de una onda esférica.

---

**Interpretación física.**

 $E_r$ decae como $1/r^3$ y $1/r^2$: nunca llega a ser radiación. $E_\theta$ contiene tres potencias, $1/r^3$ (electrostática), $1/r^2$ (inducción) y $1/r$ (radiación), y es la única componente que sobrevive a gran distancia.

---

### **6. Forma adimensional**

Se realiza un cambio de variables para evitar tratar con valores numéricos muy grandes o muy pequeños que puedan quedar atados a un valor específico y no visualizarse correctamente.

Con $\xi=kr$ (es decir $r=\xi/k$), la escala de campo $E_0\equiv\dfrac{p_0k^3}{4\pi\epsilon_0}=Ck^3$, y usando $\mu_0\omega kc=k^2/\epsilon_0$ para escribir $B$ como $cB$, se sustituye $1/r^n=k^n/\xi^n$ en cada término. Por ejemplo, para $E_r$:

$$E_r=C\cos\theta\,\frac{2k^3}{\xi^3}\Big[\cos u-\xi\sin u\Big]=E_0\,2\cos\theta\left[\frac{\cos u}{\xi^3}-\frac{\sin u}{\xi^2}\right]$$

y análogamente para las otras dos componentes:

$$\boxed{\begin{aligned}
\frac{E_r}{E_0}&=2\cos\theta\left[\frac{\cos u}{\xi^3}-\frac{\sin u}{\xi^2}\right]\\[4pt]
\frac{E_\theta}{E_0}&=\sin\theta\left[\frac{\cos u}{\xi^3}-\frac{\cos u}{\xi}-\frac{\sin u}{\xi^2}\right]\\[4pt]
\frac{cB_\phi}{E_0}&=-\sin\theta\left[\frac{\cos u}{\xi}+\frac{\sin u}{\xi^2}\right]
\end{aligned}}\qquad u=\omega t-\xi$$

**Interpretación física.** Todo el sistema depende de *una* sola variable de distancia, $\xi=2\pi r/\lambda$, de la fase $u$ y del ángulo $\theta$. Medir $r$ en unidades de $\lambda$ es lo que hace que la visualización "explore distancias respecto a la longitud de onda", como pide el enunciado.

---

### **7. Los dos regímenes**

| | Región cercana ($\xi\ll1$) | Región de radiación ($\xi\gg1$) |
|---|---|---|
| $E_r/E_0$ | $\approx 2\cos\theta\cos u/\xi^3$ | $O(1/\xi^2)$, despreciable |
| $E_\theta/E_0$ | $\approx\sin\theta\cos u/\xi^3$ | $\approx-\sin\theta\cos u/\xi$ |
| $cB_\phi/E_0$ | $\approx-\sin\theta\sin u/\xi^2$ | $\approx-\sin\theta\cos u/\xi$ |
| Fase $\mathbf E$ vs $\mathbf B$ | cuadratura ($90^\circ$) | en fase |
| Carácter | $\mathbf E$ casi electrostático, $\mathbf B$ casi magnetostático | onda esférica transversal |

El cruce entre ambos comportamientos ocurre en $\xi\sim1$, es decir $r\sim\lambda/2\pi\approx0.16\,\lambda$.

**Cómo se obtiene la tabla:** Cada componente es suma de términos con distinta potencia de $1/\xi$ y todos con coeficiente de orden 1 (a menos de $\sin\theta$ o $\cos\theta$). Si $\xi\ll1$ domina la potencia más alta de $1/\xi$: $1/\xi^3$ en $\mathbf E$ y $1/\xi^2$ en $\mathbf B$. Si $\xi\gg1$ dominan las potencias más bajas: $1/\xi$ en $E_\theta$ y en $cB_\phi$, mientras que $E_r$ (cuya mayor potencia sobreviviente es $1/\xi^2$) queda por debajo de ellos.

**Interpretación física.**

En la zona cercana $\mathbf E$ sigue a $p(t)$ y $\mathbf B$ sigue a $\dot p(t)$, así que están en cuadratura y la energía solo "va y viene" entre ellos. En la zona de radiación $E_\theta=cB_\phi$ en módulo y fase, y el par $(\mathbf E,\mathbf B)$ se comporta localmente como una onda plana.

### **Predicción previa a las visualizaciones (sección 4.1 de la guía)**

Antes de generar las figuras, a partir de la dependencia con $r$ de los términos de la sección 7, se espera:

1. **Forma de las líneas de $\mathbf E$.** Para $r\ll\lambda$ dominan los términos $1/r^3$, idénticos a los del dipolo electrostático pero con $p$ evaluado en $t-r/c\approx t$: las líneas deberían verse como las de un dipolo estático que cambia de intensidad. Para $r\gg\lambda$ solo sobrevive $E_\theta\propto1/r$, así que $\mathbf E$ debería ser transversal ($\perp\hat r$), tangente a frentes esféricos.
2. **Fase entre $\mathbf E$ y $\mathbf B$.** En la zona cercana $\mathbf E\propto p\propto\cos u$ y $\mathbf B\propto\dot p\propto\sin u$: se espera un desfase de $90^\circ$. En la zona lejana ambos términos $1/r$ van con $\cos u$: se espera que estén en fase, con $E_\theta=cB_\phi$.
3. **Flujo de energía.** Con $\mathbf E$ y $\mathbf B$ en cuadratura, $\mathbf S\propto\cos u\sin u$ cambia de signo dos veces por periodo, así que en la zona cercana esperamos que la energía vaya y venga **sin flujo neto apreciable**. En la zona lejana esperamos $S_r\propto\cos^2u\ge0$: flujo siempre saliente.
4. **Distancia de transición.** Todos los términos de cada componente tienen coeficientes de orden 1 en la variable $\xi=kr$, así que la transición debería ocurrir en $\xi\sim1$, es decir $r\sim\lambda/2\pi\approx0.16\lambda$, y no en $r\sim\lambda$.

La siguiente celda define las funciones que implementan las expresiones anteriores (en coordenadas esféricas y en el plano $xz$), y luego se comprueba numéricamente que cumplen las ecuaciones de Maxwell.

**Nota.** Esta comprobación *no* resuelve las ecuaciones de Maxwell: solo evalúa, con derivadas numéricas, que las soluciones analíticas de las secciones 4 y 5 las satisfacen. Sirve para detectar errores de álgebra o de transcripción al código.

```python
# bibliotecas numericas y graficas
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import TwoSlopeNorm
# controles interactivos (vienen instalados por defecto en Colab;
# fuera de Colab: pip install ipywidgets)
from ipywidgets import interact, FloatSlider, Checkbox, fixed

# descarga de archivos: solo existe en Google Colab. Fuera de Colab las figuras
# se guardan en la carpeta de trabajo y el notebook se ejecuta igual.
try:
    from google.colab import files
except ImportError:
    files = None

# integral trapezoidal compatible con numpy 1.x y 2.x
trapecio = getattr(np, "trapezoid", None) or np.trapz


def campos_esfericos(xi, theta, tau):
    """Er/E0, Etheta/E0 y cBphi/E0 en coordenadas esfericas.

    xi = k r = 2 pi r / lambda, theta = angulo polar, tau = t / T.
    """
    # fase retardada u = omega t - xi, con omega t = 2 pi tau
    u = 2*np.pi*tau - xi
    cu, su = np.cos(u), np.sin(u)
    # componente radial de E
    Er = 2*np.cos(theta)*(cu/xi**3 - su/xi**2)
    # componente polar de E (tres potencias de 1/xi)
    Eth = np.sin(theta)*(cu/xi**3 - cu/xi - su/xi**2)
    # campo magnetico (multiplicado por c), componente azimutal
    cBph = -np.sin(theta)*(cu/xi + su/xi**2)
    return Er, Eth, cBph

# aplica sección 8
def poynting_esfericos(xi, theta, tau):
    """S_r y S_theta en unidades de eps0 c E0^2.

    Se usa S = (1/mu0) E x B con B = B_phi phi_hat:
    S_r = E_theta B_phi / mu0,  S_theta = -E_r B_phi / mu0.
    """
    Er, Eth, cBph = campos_esfericos(xi, theta, tau)
    return Eth*cBph, -Er*cBph

# pasa los campos a coordenadas cartesianas en el plano y=0
def campos_xz(X, Z, tau):
    """Campos en el plano y = 0, con X y Z en unidades de lambda.

    Devuelve R, Ex, Ez, By, Sx, Sz, Sr. El campo B (en unidades de E0/c)
    y S (en unidades de eps0 c E0^2) se pasan a cartesianas.
    """
    # distancia al origen (se evita R = 0 exacto)
    R = np.maximum(np.hypot(X, Z), 1e-12)
    xi = 2*np.pi*R
    # cos(theta), sin(theta) y signo de x (phi = 0 si x > 0, phi = pi si x < 0)
    cos_t = np.clip(Z/R, -1.0, 1.0)
    sin_t = np.abs(X)/R
    sg = np.sign(X)
    Er, Eth, cBph = campos_esfericos(xi, np.arccos(cos_t), tau)
    # r_hat = (sin_t*sg, cos_t),  theta_hat = (cos_t*sg, -sin_t)  en (x, z)
    Ex = (Er*sin_t + Eth*cos_t)*sg
    Ez = Er*cos_t - Eth*sin_t
    # phi_hat = +y_hat para x > 0 y -y_hat para x < 0
    By = cBph*sg
    Sr, Sth = Eth*cBph, -Er*cBph
    Sx = (Sr*sin_t + Sth*cos_t)*sg
    Sz = Sr*cos_t - Sth*sin_t
    return R, Ex, Ez, By, Sx, Sz, Sr
```

```python
# se verifican las ecuaciones de Maxwell para evitar errores en los calculos,
# en puntos de las tres regiones (cercana, intermedia y de radiacion)

def verifica_maxwell(h=1e-6):
    """Residuos relativos de las ecuaciones de Maxwell en forma adimensional.

    Se usa u = 2 pi tau - xi, de modo que d/dt a xi fijo equivale a
    (omega/2pi) d/dtau. Con r = xi/k las ecuaciones quedan:
      Gauss   : (1/xi^2) d(xi^2 E_r)/dxi + (1/(xi sin th)) d(sin th E_th)/dth = 0
      Faraday : (1/xi)[d(xi E_th)/dxi - dE_r/dtheta] = -(1/2pi) d(cB_phi)/dtau
      Ampere r: (1/(xi sin th)) d(sin th cB_phi)/dtheta = (1/2pi) dE_r/dtau
      Ampere t: -(1/xi) d(xi cB_phi)/dxi = (1/2pi) dE_th/dtau
    (div B = 0 se cumple automaticamente: B = B_phi(r, theta) phi_hat.)
    """
    # (xi, theta, tau): zona cercana, intermedia y de radiacion
    puntos = [(0.2, 0.4, 0.10), (0.7, 1.2, 0.60), (1.7, 0.9, 0.23),
              (4.0, 2.0, 0.80), (25.0, 1.4, 0.37)]

    def rel(a, b):
        # residuo relativo |a - b| / max(|a|, |b|)
        return abs(a - b)/max(abs(a), abs(b), 1e-300)

    print(f"{'xi':>6} {'theta':>6} {'Gauss':>10} {'Faraday':>10} {'Ampere r':>10} {'Ampere th':>10}")
    peor = 0.0
    for xi, th, tau in puntos:
        F = campos_esfericos
        # derivadas centradas respecto a xi, theta y tau (paso relativo en xi)
        hx = h*xi
        dxi = lambda i: (F(xi+hx, th, tau)[i] - F(xi-hx, th, tau)[i])/(2*hx)
        dth = lambda i: (F(xi, th+h, tau)[i] - F(xi, th-h, tau)[i])/(2*h)
        dta = lambda i: (F(xi, th, tau+h)[i] - F(xi, th, tau-h)[i])/(2*h)
        Er, Eth, cB = F(xi, th, tau)

        # Gauss: los dos sumandos deben cancelarse
        g1 = (2*xi*Er + xi**2*dxi(0))/xi**2
        g2 = (np.cos(th)*Eth + np.sin(th)*dth(1))/(xi*np.sin(th))
        r_gauss = rel(g1, -g2)
        # d(xi E_th)/dxi = E_th + xi dE_th/dxi
        r_far = rel((Eth + xi*dxi(1) - dth(0))/xi, -dta(2)/(2*np.pi))
        # d(sin th cB)/dtheta = cos th cB + sin th dcB/dtheta
        r_amr = rel((np.cos(th)*cB + np.sin(th)*dth(2))/(xi*np.sin(th)), dta(0)/(2*np.pi))
        # d(xi cB)/dxi = cB + xi dcB/dxi
        r_amt = rel(-(cB + xi*dxi(2))/xi, dta(1)/(2*np.pi))
        peor = max(peor, r_gauss, r_far, r_amr, r_amt)
        print(f"{xi:6.1f} {np.degrees(th):5.0f}° {r_gauss:10.1e} {r_far:10.1e} {r_amr:10.1e} {r_amt:10.1e}")
    print("Residuo relativo maximo:", f"{peor:.1e}")
    print("Los campos cumplen las ecuaciones de Maxwell:", bool(peor < 1e-5))


verifica_maxwell()

# comprobacion de la transformacion a cartesianas: E x B = (-Ez By, 0, Ex By)
X, Z = np.meshgrid(np.linspace(-1, 1, 40), np.linspace(-1, 1, 40))
R, Ex, Ez, By, Sx, Sz, Sr = campos_xz(X, Z, 0.15)
print("S cartesiano consistente con E x B:",
      np.allclose(Sx, -Ez*By) and np.allclose(Sz, Ex*By))
```

### **8. Vector de Poynting, potencia y patrón angular**

Se define el vector de Poynting: $\mathbf S=\dfrac1{\mu_0}\mathbf E\times\mathbf B$

con $\mathbf B=B_\phi\hat\phi$.

Usando $\hat r\times\hat\phi=-\hat\theta$ y $\hat\theta\times\hat\phi=\hat r$:

$$\mathbf E\times\mathbf B=E_rB_\phi\,(\hat r\times\hat\phi)+E_\theta B_\phi\,(\hat\theta\times\hat\phi)=-E_rB_\phi\,\hat\theta+E_\theta B_\phi\,\hat r$$

$$S_r=\frac{E_\theta B_\phi}{\mu_0}=\epsilon_0cE_0^2\,\frac{E_\theta}{E_0}\frac{cB_\phi}{E_0},\qquad S_\theta=-\frac{E_rB_\phi}{\mu_0}=-\epsilon_0cE_0^2\,\frac{E_r}{E_0}\frac{cB_\phi}{E_0}$$

con $1/\mu_0=\epsilon_0c^2$

---

### **Promedio temporal exacto**

Sobre un periodo la fase $u=\omega t-\xi$ recorre $2\pi$ (el corrimiento $\xi$ no cambia el resultado), y

$$\langle\cos^2u\rangle=\tfrac12,\qquad\langle\sin^2u\rangle=\tfrac12,\qquad\langle\cos u\sin u\rangle=0$$

Entonces, si $f=a_f\cos u+b_f\sin u$ y $g=a_g\cos u+b_g\sin u$:

$$\langle fg\rangle=a_fa_g\langle\cos^2u\rangle+b_fb_g\langle\sin^2u\rangle+(a_fb_g+b_fa_g)\langle\cos u\sin u\rangle=\tfrac12\,(a_fa_g+b_fb_g)$$

**1. Para $S_r$**

Los coeficientes (sin el factor $\sin\theta$) son

$$E_\theta/E_0:\quad a_f=\frac1{\xi^3}-\frac1\xi,\quad b_f=-\frac1{\xi^2};\qquad\qquad cB_\phi/E_0:\quad a_g=-\frac1\xi,\quad b_g=-\frac1{\xi^2}$$

$$\left\langle\frac{E_\theta}{E_0}\frac{cB_\phi}{E_0}\right\rangle=\frac{\sin^2\theta}{2}\left[-\left(\frac1{\xi^3}-\frac1\xi\right)\frac1\xi+\frac1{\xi^2}\cdot\frac1{\xi^2}\right]=\frac{\sin^2\theta}{2}\left[-\frac1{\xi^4}+\frac1{\xi^2}+\frac1{\xi^4}\right]=\frac{\sin^2\theta}{2\xi^2}$$

Los términos $1/\xi^4$ **se cancelan exactamente**.

**2. Para $S_\theta$**

$E_r/E_0$ tiene $a_f=\dfrac2{\xi^3}\cos\theta$, $b_f=-\dfrac2{\xi^2}\cos\theta$, y $cB_\phi/E_0$ tiene los mismos $a_g$, $b_g$ de arriba (multiplicados por $\sin\theta$):

$$\langle E_r\,cB_\phi\rangle\propto\frac12\left[\left(\frac1{\xi^3}\right)\left(-\frac1\xi\right)+\left(-\frac1{\xi^2}\right)\left(-\frac1{\xi^2}\right)\right]=\frac12\left[-\frac1{\xi^4}+\frac1{\xi^4}\right]=0$$

Paso a unidades físicas

Con $\xi=kr$, $E_0=p_0k^3/4\pi\epsilon_0$, $k=\omega/c$ y $1/\epsilon_0=\mu_0c^2$:

$$\langle S_r\rangle=\epsilon_0cE_0^2\,\frac{\sin^2\theta}{2k^2r^2}=\frac{c\,p_0^2k^4}{32\pi^2\epsilon_0}\frac{\sin^2\theta}{r^2}=\frac{\mu_0p_0^2\omega^4}{32\pi^2c}\,\frac{\sin^2\theta}{r^2}$$

---

### **Potencia total**

Se integra sobre una esfera de radio $r$ ($da=r^2\,d\Omega$), usando $\int_0^\pi\sin^3\theta\,d\theta=\left[-\cos\theta+\frac{\cos^3\theta}{3}\right]_0^\pi=\frac43$:

$$P=\oint\langle\mathbf S\rangle\cdot d\mathbf a=\frac{\mu_0p_0^2\omega^4}{32\pi^2c}\int_0^{2\pi}d\phi\int_0^\pi\sin^3\theta\,d\theta=\frac{\mu_0p_0^2\omega^4}{32\pi^2c}\cdot2\pi\cdot\frac43=\frac{\mu_0p_0^2\omega^4}{12\pi c}$$

Resultado

$$\boxed{\langle\mathbf S\rangle=\frac{\mu_0p_0^2\omega^4}{32\pi^2c}\,\frac{\sin^2\theta}{r^2}\,\hat r,\qquad P=\frac{\mu_0p_0^2\omega^4}{12\pi c}}$$

---

**Interpretación física.**

Esta forma es **exacta para todo $r$** (bajo $d\ll\lambda$), no solo en la zona lejana:

- El flujo neto de energía es siempre radial y decae como $1/r^2$, de modo que la potencia que cruza cualquier esfera es la misma (conservación de la energía).
- Instantáneamente, en la zona cercana $S_r$ cambia de signo y $S_\theta\neq0$: hay energía **reactiva** que oscila entre el dipolo y el campo cercano sin escapar.
- El patrón angular de la energía radiada es $\sin^2\theta$: nula a lo largo del eje del dipolo y máxima en el plano ecuatorial.

### **Verificación numérica del promedio temporal y del flujo**

La definición del enunciado es $\langle\mathbf S\rangle=\frac1T\int_0^T\mathbf S\,dt$. En la celda siguiente esa integral se calcula numéricamente (promedio de muestras uniformes en un periodo) a partir de $\mathbf S(t)$ y se compara con el resultado analítico $\xi^2\langle S_r\rangle=\sin^2\theta/2$ y $\langle S_\theta\rangle=0$. Luego se integra $S_r$ sobre esferas de distinto radio.

Las unidades son $\epsilon_0cE_0^2$ para $\mathbf S$ y $\epsilon_0cE_0^2/k^2$ para el flujo $\Phi=\oint S_r\,r^2d\Omega$. En esas unidades la potencia media es $\langle\Phi\rangle=\frac12\cdot\frac{8\pi}{3}=\frac{4\pi}{3}$ para todo $\xi$.

```python
def promedio_temporal(xi, theta, N=4000):
    """<S_r> y <S_theta> en unidades de eps0 c E0^2, como (1/T) int S dt."""
    # N muestras uniformes en un periodo: para un polinomio trigonometrico
    # la suma de rectangulos reproduce la integral exactamente
    tau = np.arange(N)/N
    Sr, Sth = poynting_esfericos(xi, theta, tau)
    return Sr.mean(), Sth.mean()


print("Promedio temporal numerico vs. resultado analitico")
print(f"{'xi':>6} {'theta':>8} {'xi^2<Sr> (num)':>16} {'sin^2(th)/2':>12} {'xi^2<Sth> (num)':>16}")
for th in (np.pi/2, np.pi/3):
    for xi in (0.1, 0.5, 1.0, 3.0, 10.0, 30.0):
        Sr_m, Sth_m = promedio_temporal(xi, th)
        print(f"{xi:6.1f} {np.degrees(th):7.1f}° {xi**2*Sr_m:16.8f} {np.sin(th)**2/2:12.8f} {xi**2*Sth_m:16.2e}")

# ================================================================================================================ #


def flujo_esfera(xi, tau, M=2001):
    """Flujo de S a traves de la esfera de radio xi, en unidades eps0 c E0^2 / k^2.

    Phi = int S_r r^2 dOmega = 2 pi int S_r xi^2 sin(theta) dtheta (r = xi/k).
    """
    th = np.linspace(0, np.pi, M)
    Sr, _ = poynting_esfericos(xi, th, tau)
    return 2*np.pi*trapecio(Sr*xi**2*np.sin(th), th)


# integral angular analitica: int sin^2(theta) dOmega = 8 pi / 3
th = np.linspace(0, np.pi, 4001)
print("\nint sin^2(theta) dOmega =", 2*np.pi*trapecio(np.sin(th)**3, th), " (8pi/3 =", 8*np.pi/3, ")")

# potencia media a traves de esferas de distinto radio: debe ser 4 pi / 3
tau_malla = np.arange(200)/200
print("\nFlujo medio a traves de la esfera (esperado 4pi/3 =", round(4*np.pi/3, 6), ")")
for xi in (0.1, 0.5, 1.0, 3.0, 10.0, 30.0):
    phi_medio = np.mean([flujo_esfera(xi, t) for t in tau_malla])
    print(f"  xi = {xi:5.1f}   <Phi> = {phi_medio:.6f}")

# ===================================================================================================== #

fig, axs = plt.subplots(1, 2, figsize=(13, 4.5))

# (a) xi^2 S_r en el ecuador contra el tiempo, para varias distancias
tau = np.linspace(0, 1, 400)
for xi in (0.3, 1.0, 3.0, 10.0):
    Sr, _ = poynting_esfericos(xi, np.pi/2, tau)
    axs[0].plot(tau, xi**2*Sr, label=rf"$\xi={xi}$")
axs[0].axhline(0.5, color="k", ls="--", lw=1, label=r"$\xi^2\langle S_r\rangle = 1/2$")
axs[0].axhline(0, color="gray", lw=0.6)
axs[0].set_xlabel(r"$t/T$")
axs[0].set_ylabel(r"$\xi^2 S_r/(\epsilon_0 c E_0^2)$   en $\theta=\pi/2$")
axs[0].set_title("Poynting radial instantáneo (ecuador)")
# escala simlog: permite ver a la vez la zona cercana (amplitud grande) y la lejana
axs[0].set_yscale("symlog", linthresh=1)
axs[0].legend(fontsize=8)

# (b) flujo instantaneo total a traves de una esfera
for xi in (0.5, 2.0, 10.0):
    phi = np.array([flujo_esfera(xi, t, M=801) for t in tau])
    axs[1].plot(tau, phi, label=rf"$\xi={xi}$")
axs[1].axhline(4*np.pi/3, color="k", ls="--", lw=1, label=r"$\langle\Phi\rangle=4\pi/3$")
axs[1].axhline(0, color="gray", lw=0.6)
axs[1].set_xlabel(r"$t/T$")
axs[1].set_ylabel(r"$\Phi = \oint S_r\, r^2 d\Omega$  (unidades $\epsilon_0 c E_0^2/k^2$)")
axs[1].set_title("Flujo instantáneo a través de una esfera")
axs[1].set_yscale("symlog", linthresh=1)
axs[1].legend(fontsize=8)
plt.tight_layout()
plt.show()
```

**Análisis**

- La tabla confirma que $\xi^2\langle S_r\rangle=\sin^2\theta/2$ y $\langle S_\theta\rangle=0$ para **cualquier** $\xi$, desde $0.1$ hasta $30$: los términos $1/\xi^4$ se cancelan también numéricamente.
- El flujo medio a través de la esfera vale $4\pi/3$ sin importar su radio: toda la energía que sale del dipolo cruza cualquier esfera, sin acumularse ni perderse (conservación de la energía).
- Instantáneamente, en el panel izquierdo, para $\xi=0.3$ la curva toma valores negativos y mucho más grandes que el promedio: en la zona cercana la energía entra y sale del campo cercano (energía **reactiva**), y solo una pequeña fracción neta escapa. Para $\xi\gtrsim3$ la parte negativa es despreciable (menor que $10^{-3}$) y la curva oscila como $\cos^2u$ entre $0$ y $1$, con promedio $1/2$: el flujo es esencialmente saliente.
- El panel derecho muestra lo mismo para el flujo total: en zona de radiación $\Phi(t)=\frac{8\pi}{3}\cos^2u\ge0$.

- `Poynting`: muestra u oculta las flechas.

**Cómo usarlo:** Empezar con $L\approx0.03$ (zona cercana, $\xi\lesssim0.2$), subir a $L\approx0.4$ (transición, contiene $r\sim\lambda/2\pi$) y luego a $L\approx3$ (radiación), moviendo `t/T` en cada caso.



```python
def mapa(L=1.0, tau=0.0, flechas_S=True, puntos=(), circulo=None):
    """Mapas de E, cB_y y S en el plano xz, ventana [-L, L]^2 (en lambdas).

    puntos: lista opcional de (x, z, etiqueta) para marcar en ambos paneles.
    circulo: radio (en lambdas) de una circunferencia de referencia, o None.
    """
    # malla con numero par de puntos: nunca cae exactamente en x = 0
    n = 300
    x = np.linspace(-L, L, n)
    z = np.linspace(-L, L, n)
    X, Z = np.meshgrid(x, z)
    R, Ex, Ez, By, Sx, Sz, Sr = campos_xz(X, Z, tau)

    # se excluye el entorno de la singularidad del dipolo puntual
    mask = R < 0.04*L
    Ex = np.ma.masked_where(mask, Ex)
    Ez = np.ma.masked_where(mask, Ez)
    By = np.ma.masked_where(mask, By)

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13.5, 6))

    # ---------- panel izquierdo: E (lineas) y cB_y (color) ----------
    # escala de color simetrica, saturada para ver estructura lejos de la fuente
    vmax = np.percentile(np.abs(By.compressed()), 90)
    pc1 = ax1.pcolormesh(x, z, By, norm=TwoSlopeNorm(0, -vmax, vmax),
                         cmap="RdBu_r", shading="auto")
    # lineas de campo electrico
    ax1.streamplot(x, z, Ex, Ez, color="k", density=1.6, linewidth=0.7, arrowsize=0.8)
    ax1.set_title(rf"$\mathbf{{E}}$ (líneas) y $cB_y$ (color) — $\pm{L:.2f}\,\lambda$, $t={tau:.2f}\,T$")
    fig.colorbar(pc1, ax=ax1, label=r"$cB_y/E_0$ (saturado)", shrink=0.8)

    # ---------- panel derecho: flujo de energia ----------
    # xi^2 S_r: en la zona de radiacion tiende a sin^2(th) cos^2(u) >= 0
    # y en la zona cercana cambia de signo (energia reactiva)
    flujo = np.ma.masked_where(mask, Sr*(2*np.pi*R)**2)
    vS = np.percentile(np.abs(flujo.compressed()), 90)
    pc2 = ax2.pcolormesh(x, z, flujo, norm=TwoSlopeNorm(0, -vS, vS),
                         cmap="PiYG", shading="auto")
    if flechas_S:
        # flechas de Poynting en una malla gruesa, normalizadas (solo direccion)
        s = 18
        mod = np.hypot(Sx, Sz) + 1e-30
        ux = np.where(mask, np.nan, Sx/mod)
        uz = np.where(mask, np.nan, Sz/mod)
        ax2.quiver(X[::s, ::s], Z[::s, ::s], ux[::s, ::s], uz[::s, ::s],
                   color="k", scale=30, width=0.004)
    ax2.set_title(r"Flujo de energía: $\xi^2 S_r$ (color) y dirección de $\mathbf{S}$ (flechas)")
    fig.colorbar(pc2, ax=ax2, label=r"$\xi^2 S_r/(\epsilon_0 c E_0^2)$ (saturado)", shrink=0.8)

    for ax in (ax1, ax2):
        # puntos elegidos para el analisis (solo los que caen dentro de la ventana)
        for xp, zp, etq in puntos:
            if abs(xp) <= L and abs(zp) <= L:
                ax.plot(xp, zp, marker="o", ms=9, mfc="gold", mec="k", zorder=5)
                ax.annotate(etq, (xp, zp), xytext=(6, 6), textcoords="offset points",
                            fontsize=11, fontweight="bold", color="k",
                            bbox=dict(boxstyle="round,pad=0.15", fc="gold", alpha=0.8))
        if circulo is not None:
            ax.add_patch(plt.Circle((0, 0), circulo, fill=False, ls="--", lw=1.5, color="dimgray"))
        ax.set_aspect("equal")
        ax.set_xlabel(r"$x/\lambda$")
        ax.set_ylabel(r"$z/\lambda$")
    plt.tight_layout()
    plt.show()


# controles interactivos: tamano de ventana respecto a lambda y tiempo
interact(mapa,
         L=FloatSlider(value=1.0, min=0.01, max=4.0, step=0.01, description="L/λ"),
         tau=FloatSlider(value=0.0, min=0.0, max=1.0, step=0.05, description="t/T"),
         flechas_S=Checkbox(value=True, description="Poynting"),
         puntos=fixed(()), circulo=fixed(None));
```

**Mapas fijos para su análisis**

Se eligen ventanas representativas de los tres regímenes que pide la guía ($r\ll\lambda$, $r\sim\lambda/2\pi$ y $r\gg\lambda$). Como en el borde de la ventana $\xi=2\pi L$:

| Ventana | $\xi$ en el borde | Régimen | Cociente $\dfrac{1/\xi^3}{1/\xi}=\dfrac1{\xi^2}$ en el borde |
|---|---|---|---|
| $L=0.03\,\lambda$ | $\approx0.19$ | **zona cercana** | $\approx28$: domina el término electrostático |
| $L=0.4\,\lambda$ | $\approx2.5$ | **transición**: contiene $r\sim\lambda/2\pi\approx0.16\lambda$ (circunferencia punteada) | $\approx0.16$ en el borde y $1$ sobre la circunferencia: los tres términos son comparables |
| $L=3\,\lambda$ | $\approx19$ | **zona de radiación** | $\approx3\times10^{-3}$: domina la radiación |

La zona cercana se muestra en **dos instantes** ($t=0.15\,T$ y $t=0.40\,T$) porque el carácter reactivo del flujo de energía solo se aprecia comparando instantes. Los otros dos regímenes se muestran en $t=0.15\,T$.

Los puntos $P_1$ (zona cercana) y $P_2$ (zona de radiación), marcados en amarillo, se usan en la sección 4.3. Ambos están en la dirección $\theta=60^\circ$.

```python
# puntos usados en la seccion 4.3 (misma direccion theta = 60 grados)
TH_P = np.deg2rad(60)
R_P1, R_P2 = 0.02, 2.5          # distancias en unidades de lambda
P1 = (R_P1*np.sin(TH_P), R_P1*np.cos(TH_P), "$P_1$")
P2 = (R_P2*np.sin(TH_P), R_P2*np.cos(TH_P), "$P_2$")

# zona cercana en dos instantes (energia que va y viene)
for tau in (0.15, 0.40):
    mapa(L=0.03, tau=tau, puntos=[P1])
# transicion y zona de radiacion
mapa(L=0.4, tau=0.15, circulo=1/(2*np.pi))
mapa(L=3.0, tau=0.15, puntos=[P2])
```

**Interpretación de los mapas.**

- **Ventana cercana** ($L=0.03\lambda$, $\xi\lesssim0.2$). Las líneas de $\mathbf E$ son las del dipolo electrostático: salen del polo superior y llegan al inferior. El campo magnético forma anillos alrededor del eje (color antisimétrico respecto a $x=0$). En el panel derecho, el signo de $\xi^2S_r$ **se invierte** entre $t=0.15\,T$ y $t=0.40\,T$, y las flechas de Poynting cambian de sentido: la energía sale y regresa al dipolo (energía reactiva).
- **Ventana de transición** ($L=0.4\lambda$; la circunferencia punteada es $r=\lambda/2\pi$, $\xi=1$). Las líneas de $\mathbf E$ empiezan a desprenderse del dipolo y a cerrarse sobre sí mismas, formando lazos. Aparece la primera capa de flujo saliente en el borde, mientras que cerca del origen sigue habiendo flujo de ambos signos.
- **Ventana lejana** ($L=3\lambda$, $\xi\gg1$). Los frentes de onda son casi esferas concéntricas. $\mathbf E$ es tangente a ellas (transversal a $\hat r$), $\mathbf B$ es perpendicular al plano y las capas de $cB_y$ se alternan con periodo $\lambda$. El color de $\xi^2S_r$ es siempre $\ge0$, modulado por $\sin^2\theta\cos^2u$: las capas verdes son los frentes de energía que se alejan, y son nulas sobre el eje. Las flechas de Poynting apuntan radialmente hacia afuera, y $\mathbf E\times\mathbf B\parallel\hat r$.

### **4.1 Región cercana y región de radiación: características que cambian**

Se identifican tres características concretas, usando las figuras anteriores.

#### **Característica 1: la forma de las líneas de $\mathbf E$**

1. **Dónde se observa.** Panel izquierdo de los mapas. En $L=0.03\lambda$ las líneas van del polo superior al inferior, como en el dipolo estático. En $L=0.4\lambda$, por fuera de la circunferencia $r=\lambda/2\pi$, aparecen lazos cerrados desprendidos del dipolo a ambos lados. En $L=3\lambda$ las líneas son tangentes a los frentes de onda y se cierran formando lazos de tamaño $\sim\lambda/2$.
2. **Términos responsables.** Cerca: los términos $\cos u/\xi^3$ de $E_r$ y $E_\theta$, que son el campo del dipolo electrostático evaluado en $t-r/c$. Lejos: el término $-\sin\theta\cos u/\xi$ de $E_\theta$, que es transversal; el término $E_r\propto1/\xi^2$, pequeño pero no nulo, es el que cierra los lazos cerca del eje.
3. **Explicación física.** Para $r\ll\lambda$ el retardo es despreciable y el campo sigue instantáneamente a $p(t)$, terminando en las cargas. Para $r\gg\lambda$ el campo ya no está ligado a las cargas: como $\nabla\cdot\mathbf E=0$ fuera de la fuente, las líneas que no terminan en cargas deben cerrarse sobre sí mismas, y esos lazos viajan hacia afuera.
4. **¿Coincidió con la predicción?** Parcialmente. Predijimos el comportamiento dipolar cerca y transversal lejos, pero no anticipamos los **lazos cerrados**. Tuvimos que agregar al razonamiento que la ley de Gauss sin fuentes obliga a que las líneas desprendidas se cierren.

#### **Característica 2: la fase entre $\mathbf E$ y $\mathbf B$**

1. **Dónde se observa.** Figura de campos a tiempo fijo: en el panel (c) el desfase pasa de $90^\circ$ a $0^\circ$. En el panel (b) las curvas $\xi E_\theta$ y $\xi\,cB_\phi$ solo coinciden para $\xi\gtrsim5$.
2. **Términos responsables.** Cerca: $E_\theta\approx\sin\theta\cos u/\xi^3$ y $cB_\phi\approx-\sin\theta\sin u/\xi^2$ (cuadratura). Lejos: los términos $-\sin\theta\cos u/\xi$ de ambos (en fase).
3. **Explicación física.** El campo eléctrico cercano lo produce la carga, $\propto p$, y el magnético lo produce la corriente, $\propto\dot p$, que está adelantada un cuarto de periodo. En la zona de radiación ambos campos se generan mutuamente (inducción de Faraday y corriente de desplazamiento) y forman una onda autosostenida, que exige $E=cB$ en fase.
4. **¿Coincidió con la predicción?** Sí. Además, el panel (c) confirma cuantitativamente que la transición está en $\xi\sim1$ (desfase de $45^\circ$), es decir $r\approx0.16\lambda$ y no $r\sim\lambda$.

#### **Característica 3: el sentido del flujo de energía**

1. **Dónde se observa.** Panel derecho de los mapas y figura del vector de Poynting en el tiempo. En $L=0.03\lambda$ el color cambia de signo entre $t=0.15\,T$ y $t=0.40\,T$; para $\xi=0.3$, $\xi^2S_r$ oscila entre valores positivos y negativos grandes. En $L=3\lambda$ el color es siempre verde ($S_r\ge0$).
2. **Términos responsables.** Cerca: el producto de los términos en cuadratura, $E_\theta\,cB_\phi\propto-\cos u\sin u/\xi^5$, de promedio nulo. Lejos: el producto de los términos $1/\xi$, $\propto\cos^2u/\xi^2\ge0$.
3. **Explicación física.** Cerca, la energía se almacena en el campo y regresa al dipolo cada medio periodo (energía reactiva). Lejos, la energía se aleja y no regresa: es radiación.
4. **¿Coincidió con la predicción?** **No del todo.** Predijimos que en la zona cercana no habría flujo neto apreciable. La tabla de promedios muestra que $\xi^2\langle S_r\rangle=\sin^2\theta/2$ es **el mismo para cualquier $\xi$**, incluso en $\xi=0.1$: el flujo neto existe también muy cerca del dipolo, solo que queda oculto bajo una oscilación reactiva mucho mayor. Tuvimos que corregir la idea de que "cuadratura" significa "flujo neto cero": los términos $1/\xi^4$ se cancelan exactamente en el promedio, pero el término $1/\xi^2$ sobrevive a toda distancia (sección 4.3).

### **Distribución Angular de la Potencia Radiada**

La potencia que atraviesa un elemento de ángulo sólido $d\Omega$ a distancia $r$ es $\langle\mathbf S\rangle\cdot\hat r\,r^2\,d\Omega$, de modo que

$$\left\langle\frac{dP}{d\Omega}\right\rangle=\langle\mathbf S\rangle\cdot\hat r\;r^{2}=\frac{\mu_0p_0^{2}\omega^{4}}{32\pi^{2}c}\,\sin^{2}\theta$$

Este resultado **no depende de $r$** (el factor $1/r^2$ de $\langle S_r\rangle$ lo cancela $r^2$), por eso describe la potencia que llega "al infinito" en cada dirección. En términos del valor máximo emitido en el ecuador, $\theta=\pi/2$,

$$\left(\frac{dP}{d\Omega}\right)_{\max}=\frac{\mu_0p_0^{2}\omega^{4}}{32\pi^{2}c}
\qquad\Longrightarrow\qquad
\left\langle\frac{dP}{d\Omega}\right\rangle=\left(\frac{dP}{d\Omega}\right)_{\max}\sin^{2}\theta$$

**Consistencia con la potencia total.** $\;P=\displaystyle\int\left\langle\frac{dP}{d\Omega}\right\rangle d\Omega=\left(\frac{dP}{d\Omega}\right)_{\max}\int\sin^2\theta\,d\Omega=\left(\frac{dP}{d\Omega}\right)_{\max}\cdot\frac{8\pi}{3}=\frac{\mu_0p_0^2\omega^4}{12\pi c}$, que coincide con la sección 8.

### **Relación entre el Patrón de Radiación y la Orientación del Dipolo**

1. **Alineación con el eje del dipolo:** El dipolo está orientado a lo largo del eje $z$ ($\theta = 0^\circ$ y $\theta = 180^\circ$). El ángulo $\theta$ representa la desviación polar respecto a la línea de oscilación de las cargas.

2. **Cero de radiación en los polos** ($\theta = 0, \pi$): Dado que $\sin(0) = \sin(\pi) = 0$, la potencia radiada a lo largo del eje del dipolo es exactamente nula ($\dfrac{dP}{d\Omega} = 0$). Físicamente, las cargas aceleradas no emiten radiación a lo largo de la línea de su propia aceleración (el campo de radiación es transversal y su amplitud es proporcional a la componente de la aceleración perpendicular a $\hat r$).

3. **Máximo de radiación en el ecuador** ($\theta = \pi/2$): Dado que $\sin(\pi/2) = 1$, la máxima emisión de energía ocurre en el plano transversal al dipolo (plano $xy$).

4. **Simetría azimutal:** La expresión no depende de la coordenada azimutal $\phi$. El patrón tridimensional posee simetría de revolución (forma de "dona" o toroide), mientras que su corte bidimensional en cualquier plano meridiano (como el plano $xz$) forma un "8" perfecto tangente al origen.

5. **Patrón ligado a la dirección de $\mathbf p$.** En el problema el único vector especial es $\hat{\mathbf p}$, así que la potencia solo puede depender del ángulo $\Theta$ entre $\hat{\mathbf p}$ y $\hat r$. Si $\hat{\mathbf p}=\hat z$ entonces $\cos\Theta=\cos\theta$ y

$$\sin^2\theta=1-(\hat{\mathbf z}\cdot\hat r)^2\quad\longrightarrow\quad\left\langle\frac{dP}{d\Omega}\right\rangle=\left(\frac{dP}{d\Omega}\right)_{\max}\Big[1-(\hat{\mathbf p}\cdot\hat r)^2\Big]=\left(\frac{dP}{d\Omega}\right)_{\max}\big|\hat{\mathbf p}\times\hat r\big|^2$$

para un dipolo orientado en cualquier dirección: **el patrón gira rígidamente junto con $\mathbf p$**, y su eje de revolución siempre es el eje del dipolo.

### **Visualización**

Se presenta la gráfica con un corte meridiano (polar) junto con la superficie tridimensional para $\mathbf p\parallel\hat z$.

```python
def patron_polar(ax, alpha_deg=0.0, titulo=None):
    """Patron dP/dOmega normalizado en el plano que contiene a p y a z.

    alpha_deg es el angulo de p respecto a z (medido de z hacia x).
    """
    alpha = np.deg2rad(alpha_deg)
    theta = np.linspace(0, 2*np.pi, 721)
    # dP/dOmega / (dP/dOmega)_max = sin^2 del angulo entre p y r:
    # sin^2(Theta) = 1 - cos^2(theta - alpha)
    patron = np.sin(theta - alpha)**2

    ax.plot(theta, patron, color="crimson", lw=2.5,
            label=r"$\frac{dP/d\Omega}{(dP/d\Omega)_{\max}}=\sin^2\Theta$")
    ax.fill(theta, patron, color="crimson", alpha=0.15)

    # z arriba (theta = 0) y angulo creciente en sentido horario
    ax.set_theta_zero_location("N")
    ax.set_theta_direction(-1)
    ax.set_rlim(0, 1.5)
    ax.set_rticks([0.25, 0.5, 0.75, 1.0])
    # etiquetas radiales cerca del eje del dipolo, donde el patron casi se anula
    ax.set_rlabel_position(np.degrees(alpha) + 12)

    # eje del dipolo: linea con flecha en ambos extremos (p oscila en ambos sentidos)
    ax.annotate("", xy=(alpha, 1.25), xytext=(alpha + np.pi, 1.25),
                arrowprops=dict(arrowstyle="<->", color="black", lw=2))
    ax.text(alpha - 0.22, 1.32, r"eje de $\mathbf{p}$", ha="center", va="center",
            fontsize=10, fontweight="bold",
            bbox=dict(boxstyle="round,pad=0.15", fc="white", ec="none", alpha=0.85))

    # maximo de emision: perpendicular al eje del dipolo
    ax.annotate("máximo", xy=(alpha + np.pi/2, 1.0), xytext=(alpha + np.pi/2 + 0.45, 1.25),
                arrowprops=dict(arrowstyle="->", color="blue", lw=1.5),
                fontsize=9, color="blue")
    if titulo:
        ax.set_title(titulo, fontsize=12, fontweight="bold", pad=22)


fig = plt.figure(figsize=(13, 6))

# patron polar (corte en un plano meridiano)
ax1 = fig.add_subplot(121, projection="polar")
patron_polar(ax1, 0.0, r"Corte meridiano de $dP/d\Omega$")
ax1.legend(loc="lower right", fontsize=9, bbox_to_anchor=(1.15, -0.1))

# superficie 3D r = sin^2(theta): "dona" con simetria de revolucion
ax2 = fig.add_subplot(122, projection="3d")
th3, ph3 = np.meshgrid(np.linspace(0, np.pi, 80), np.linspace(0, 2*np.pi, 120))
rad = np.sin(th3)**2
ax2.plot_surface(rad*np.sin(th3)*np.cos(ph3), rad*np.sin(th3)*np.sin(ph3), rad*np.cos(th3),
                 facecolors=plt.cm.magma(rad), rstride=2, cstride=2, linewidth=0, antialiased=True)
# eje del dipolo (z)
ax2.plot([0, 0], [0, 0], [-1.3, 1.3], color="k", lw=2)
ax2.text(0, 0, 1.4, r"$\mathbf{p}$", fontsize=12, ha="center")
ax2.set_box_aspect((1, 1, 1))
ax2.set_xlim(-1, 1); ax2.set_ylim(-1, 1); ax2.set_zlim(-1.3, 1.3)
ax2.set_xlabel("x"); ax2.set_ylabel("y"); ax2.set_zlabel("z")
ax2.set_title(r"Superficie $r\propto dP/d\Omega=\sin^2\theta$", fontsize=12, fontweight="bold")
plt.tight_layout()
plt.show()
```

Luego el mismo patrón para dipolos inclinados, que ilustra el punto 5.

```python
# el patron depende solo del angulo entre p y r: se gira junto con el dipolo
fig = plt.figure(figsize=(15, 5.2))
for i, alpha in enumerate((0.0, 35.0, 90.0)):
    ax = fig.add_subplot(1, 3, i + 1, projection="polar")
    patron_polar(ax, alpha, rf"$\mathbf{{p}}$ a {alpha:.0f}° de $z$")
plt.tight_layout()
plt.show()

# la potencia total no depende de la orientacion: int sin^2(Theta) dOmega = 8 pi / 3
th = np.linspace(0, np.pi, 1501)
ph = np.linspace(0, 2*np.pi, 1501)
TH, PH = np.meshgrid(th, ph, indexing="ij")
for alpha_deg in (0, 35, 90):
    a = np.deg2rad(alpha_deg)
    # p_hat = (sin a, 0, cos a) y r_hat = (sin th cos ph, sin th sin ph, cos th)
    p_dot_r = np.sin(a)*np.sin(TH)*np.cos(PH) + np.cos(a)*np.cos(TH)
    integrando = (1 - p_dot_r**2)*np.sin(TH)
    I = trapecio(trapecio(integrando, ph, axis=1), th)
    print(f"alpha = {alpha_deg:3d}°   int sin^2(Theta) dOmega = {I:.6f}   (8pi/3 = {8*np.pi/3:.6f})")
```

**Análisis de Resultados**

- **Patrón.** El gráfico polar relaciona la distancia al origen con el módulo de la potencia emitida por unidad de ángulo sólido en la dirección definida por el ángulo polar $\theta$. El sistema no emite de forma isotrópica: la distribución espacial de potencia, $\sin^2\theta$, queda completamente condicionada por la dirección del vector momento dipolar $\mathbf p(t)$.
- **Orientación.** Los tres paneles inclinados muestran que, al rotar $\mathbf p$ un ángulo $\alpha$, el "8" rota el mismo ángulo: los nulos siempre están sobre el eje de $\mathbf p$ y los máximos siempre en el plano perpendicular a él. La potencia total no cambia (la integral numérica vale $8\pi/3$ en los tres casos), solo cambia hacia dónde se dirige.
- **Cercano vs. lejano.** A distancias comparables o menores que $\lambda/2\pi$ el campo está dominado por términos $1/r^3$ y $1/r^2$, $\mathbf E$ y $\mathbf B$ están en cuadratura y el flujo de energía es mayormente reactivo. Para $r\gg\lambda$ sobreviven solo los términos $1/r$: $\mathbf E$ y $\mathbf B$ están en fase con $E_\theta=cB_\phi$, y el flujo de energía es radial y con el patrón $\sin^2\theta$.
- **Conservación de la energía.** Aunque en la zona cercana el flujo instantáneo es grande y cambia de signo, el promedio temporal a través de cualquier esfera es siempre el mismo, $P=\mu_0p_0^2\omega^4/(12\pi c)$.

### **4.2 Patrón angular de radiación: valores en direcciones específicas**

Se calcula $dP/d\Omega=r^2\langle S_r\rangle$ promediando numéricamente $S(t)$ en la zona de radiación ($\xi=200$, es decir $r\approx32\lambda$), en varias direcciones, y se normaliza respecto al ecuador.

```python
# direcciones elegidas (grados respecto al eje del dipolo)
direcciones = [0, 15, 30, 45, 60, 75, 90, 120, 150]
xi_lejos = 200.0
ref = xi_lejos**2*promedio_temporal(xi_lejos, np.pi/2)[0]

print(f"{'theta':>6} {'(dP/dOmega)/(dP/dOmega)_max':>30} {'sin^2(theta)':>13}")
valores = []
for th_deg in direcciones:
    th = np.deg2rad(th_deg)
    # r^2 <S_r> en unidades de eps0 c E0^2 / k^2, normalizado al ecuador
    rel = xi_lejos**2*promedio_temporal(xi_lejos, th)[0]/ref
    valores.append(rel)
    print(f"{th_deg:5d}° {rel:30.4f} {np.sin(th)**2:13.4f}")

# patron polar con las direcciones elegidas marcadas
fig = plt.figure(figsize=(5.8, 5.8))
ax = fig.add_subplot(111, projection="polar")
patron_polar(ax, 0.0, r"Valores de $dP/d\Omega$ en direcciones elegidas")
# etiquetas radiales del lado izquierdo para no tapar los valores
ax.set_rlabel_position(-20)
for th_deg, v in zip(direcciones, valores):
    ax.plot(np.deg2rad(th_deg), v, "o", ms=7, mfc="gold", mec="k", zorder=5)
    if th_deg >= 30:
        ax.annotate(f"{v:.2f}", (np.deg2rad(th_deg), v), xytext=(7, -4),
                    textcoords="offset points", fontsize=9)
plt.tight_layout()
plt.show()
```

|---|---|
| $\theta=0^\circ$ (eje del dipolo) | $0$ |
| $\theta=30^\circ$ | $0.25$ |
| $\theta=45^\circ$ | $0.50$ |
| $\theta=60^\circ$ | $0.75$ |
| $\theta=90^\circ$ (ecuador) | $1$ |

- **Anisotropía.** A $30^\circ$ del eje se emite solo la cuarta parte que en el ecuador; a $45^\circ$, la mitad; y sobre el eje, nada. La potencia está concentrada cerca del plano ecuatorial: la mitad de la potencia total sale entre $\theta\approx70^\circ$ y $110^\circ$, una franja de solo $\pm20^\circ$ alrededor del ecuador.
- **Simetría.** Los valores para $\theta$ y $180^\circ-\theta$ son iguales (por ejemplo, $60^\circ$ y $120^\circ$ dan $0.75$): el patrón es simétrico respecto al plano ecuatorial.
- **Qué produce la dependencia angular.** En la zona de radiación solo sobrevive el término $E_\theta=-E_0\sin\theta\cos u/\xi$, que viene de $-\partial\mathbf A/\partial t$ y del término $k\sin u/r$ de $V$. Equivalentemente, $\mathbf E_{\rm rad}\propto\hat r\times(\hat r\times\ddot{\mathbf p})$, cuya magnitud es $|\ddot{\mathbf p}|\sin\theta$: **solo la componente de la aceleración de las cargas perpendicular a la línea de visión radia**. Como $\langle S_r\rangle\propto E_\theta\,cB_\phi\propto\sin^2\theta$, la potencia va como $\sin^2\theta$. Sobre el eje, $\ddot{\mathbf p}\parallel\hat r$ y su componente perpendicular es nula.

### **4.3 Transporte de energía: comparación entre dos puntos**

Se eligen los puntos marcados en los mapas, ambos en la dirección $\theta=60^\circ$:

- $P_1$: $r=0.02\,\lambda$ ($\xi\approx0.13$), en la **zona cercana** (mapa de $L=0.03\lambda$).
- $P_2$: $r=2.5\,\lambda$ ($\xi\approx15.7$), en la **zona de radiación** (mapa de $L=3\lambda$).

En cada punto se calcula $\langle\mathbf S\rangle=\frac1T\int_0^T\mathbf S\,dt$ numéricamente, su magnitud y su orientación, y también el rango de $\mathbf S(t)$ instantáneo.

```python
def analisis_punto(r_lambda, theta, N=4000):
    """<S> y rango de S(t) en un punto (unidades de eps0 c E0^2)."""
    xi = 2*np.pi*r_lambda
    tau = np.arange(N)/N
    Sr, Sth = poynting_esfericos(xi, theta, tau)
    Sr_m, Sth_m = Sr.mean(), Sth.mean()
    return dict(xi=xi, Sr=Sr_m, Sth=Sth_m, mod=np.hypot(Sr_m, Sth_m),
                ang=np.degrees(np.arctan2(Sth_m, Sr_m)), xi2Sr=xi**2*Sr_m,
                Sr_min=Sr.min()/Sr_m, Sr_max=Sr.max()/Sr_m, Sth_max=np.abs(Sth).max()/Sr_m,
                tau=tau, Sr_t=Sr, Sth_t=Sth)


res = {nombre: analisis_punto(r, TH_P) for nombre, r in (("P1", R_P1), ("P2", R_P2))}

filas = [("xi = k r", "xi", "{:.3f}"),
         ("<S_r>  [eps0 c E0^2]", "Sr", "{:.4e}"),
         ("<S_theta>  [eps0 c E0^2]", "Sth", "{:.1e}"),
         ("angulo entre <S> y r_hat", "ang", "{:.1e} grados"),
         ("xi^2 <S_r>  (= r^2<S_r> k^2)", "xi2Sr", "{:.4f}"),
         ("min S_r(t) / <S_r>", "Sr_min", "{:.1f}"),
         ("max S_r(t) / <S_r>", "Sr_max", "{:.1f}"),
         ("max |S_theta(t)| / <S_r>", "Sth_max", "{:.1f}")]
print(f"{'':32} {'P1 (cercano)':>18} {'P2 (radiacion)':>18}")
for etq, k, fmt in filas:
    print(f"{etq:32} {fmt.format(res['P1'][k]):>18} {fmt.format(res['P2'][k]):>18}")
print(f"\n<S>(P1) / <S>(P2) = {res['P1']['Sr']/res['P2']['Sr']:.1f}"
      f"   (esperado (r2/r1)^2 = {(R_P2/R_P1)**2:.1f})")

# S(t) normalizado por <S_r> en ambos puntos
fig, axs = plt.subplots(1, 2, figsize=(13, 4.3), sharex=True)
for ax, nombre in zip(axs, ("P1", "P2")):
    d = res[nombre]
    ax.plot(d["tau"], d["Sr_t"]/d["Sr"], label=r"$S_r(t)/\langle S_r\rangle$")
    ax.plot(d["tau"], d["Sth_t"]/d["Sr"], label=r"$S_\theta(t)/\langle S_r\rangle$")
    ax.axhline(1, color="k", ls="--", lw=1, label=r"$\langle S_r\rangle$")
    ax.axhline(0, color="gray", lw=0.6)
    ax.set_xlabel(r"$t/T$")
    ax.set_title(rf"${nombre[0]}_{nombre[1]}$: $r={d['xi']/(2*np.pi):.2f}\lambda$, $\theta=60^\circ$")
    ax.legend(fontsize=9)
axs[0].set_yscale("symlog", linthresh=1)
plt.tight_layout()
plt.show()
```

**Análisis**

- **Orientación.** En ambos puntos $\langle S_\theta\rangle=0$ (a precisión de máquina) y $\langle S_r\rangle>0$: el vector de Poynting promedio es **radial y hacia afuera** tanto en la zona cercana como en la de radiación.
- **Magnitud.** $\langle S_r\rangle$ es unas $1.6\times10^4$ veces mayor en $P_1$ que en $P_2$, exactamente $(r_2/r_1)^2=125^2$. Pero $r^2\langle S_r\rangle$ (en la tabla, $\xi^2\langle S_r\rangle=\sin^2 60^\circ/2=0.375$) es **igual en los dos puntos**: la potencia por unidad de ángulo sólido que cruza la esfera de radio $r_1$ es la misma que cruza la de radio $r_2$. La energía no se acumula ni se pierde entre ambas esferas; simplemente se reparte sobre un área que crece como $r^2$.
- **Lo que distingue a las dos regiones es el flujo instantáneo.** En $P_1$, $S_r(t)$ oscila entre valores cientos de veces mayores que su promedio, con signo positivo y negativo, y $S_\theta(t)$ también es grande. La energía que fluye en cada instante es casi toda reactiva y regresa al dipolo. En $P_2$, $S_r(t)$ oscila entre $\approx0$ y $\approx2\langle S_r\rangle$ (como $2\cos^2u$), sin cambiar de signo, y $S_\theta$ es pequeño: su máximo es $\approx0.1\,\langle S_r\rangle$ y, respecto a $S_r$, decrece como $1/r$.
- **Evidencia sobre el transporte de energía.** El dipolo emite energía neta de forma radial desde muy cerca de la fuente, y la misma potencia atraviesa cualquier esfera. La zona cercana no "produce" la radiación en un radio especial: lo que cambia con la distancia es que la energía reactiva (términos $1/r^4$ y $1/r^5$ de $\mathbf S$) decae mucho más rápido que la radiada ($1/r^2$), y por eso a gran distancia solo queda el flujo saliente.

## **5. Retrospectiva**

**Qué observamos.** En la tabla de promedios temporales y en el punto $P_1$, el flujo neto de energía $\xi^2\langle S_r\rangle$ es el mismo en $\xi=0.1$ que en $\xi=30$, aunque en la figura de $S_r(t)$ la curva de $\xi=0.3$ oscila con amplitudes enormes y cambia de signo.

**Qué parte de las ecuaciones lo describe.** El producto $E_\theta\,cB_\phi$ contiene términos de orden $1/\xi^5$, $1/\xi^4$, $1/\xi^3$ y $1/\xi^2$. En el promedio temporal los términos en cuadratura ($\cos u\sin u$) se anulan y los de $1/\xi^4$ se cancelan exactamente entre sí; solo sobrevive $\sin^2\theta/(2\xi^2)$.

**Interpretación física.** Cerca del dipolo coexisten dos flujos: uno reactivo, grande, que va y viene, y uno radiativo, pequeño, que siempre sale. La radiación no "nace" a una distancia particular; lo que ocurre es que el flujo reactivo decae más rápido y deja de ocultar al radiativo.

**Por qué la visualización ayudó.** Al mirar solo las fórmulas de la zona cercana ($\mathbf E\propto p$, $\mathbf B\propto\dot p$, en cuadratura) concluimos que no había flujo neto allí. Ver en la misma figura la curva de $\xi=0.3$ oscilando alrededor de la recta $\langle S_r\rangle=1/2$, y la tabla con el mismo valor para todo $\xi$, nos obligó a revisar esa conclusión.
